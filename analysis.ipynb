## 3. Load all files

```python
def load_json(path):
    rows = []
    with open(path, encoding='utf-8', errors='replace') as f:
        for line in f:
            line = line.strip()
            if line.endswith(',}'):
                line = line[:-2] + '}'
            rows.append(json.loads(line))

    df = pd.DataFrame(rows)
    df['domain'] = Path(path).parts[1]
    df['subdomain'] = Path(path).parts[2]
    df['source'] = 'json'
    return df
```

## 6. Text files

```python
text_only = []

for folder_path in [folder, folder_hd]:
    for p in folder_path.rglob("*.access.*log"):
        if '.access.json' in p.name:
            continue
        twin = p.with_name(p.name.replace('.access.', '.access.json.'))
        
        if not twin.exists(): # checking, if json already exists
            text_only.append(p) # bo'lmasa arrayga qo'shiladi

for p in text_only:
    print(p)
len(text_only)
```

```python
LINE_RE = re.compile(
    r'(?P<client_address>\S+) (?P<request_id>\S+) (?P<user>\S+) '
    r'\[(?P<datetime>[^\]]+)\] '
    r'"(?P<request>[^"]*)" '
    r'(?P<status_code>\d{3}) (?P<response_body_size>\d+) '
    r'"(?P<http_referrer>[^"]*)" '
    r'"(?P<http_user_agent>[^"]*)" '
    r'(?P<response_time>\S+)'
)
```

```python
def load_text(path):
    rows, bad = [], []  # bad for inconsistent lines
    with open(path, encoding='utf-8', errors='replace') as f:
        for line in f:
            line = line.strip()
            m = LINE_RE.fullmatch(line)
            if m:
                rows.append(m.groupdict())
            else:
                bad.append(line)

    if bad:
        print(f'{path.name}: {len(bad)} bad lines')

    df = pd.DataFrame(rows)
    df['domain'] = path.parts[1]
    df['subdomain'] = path.parts[2]
    df['host'] = path.name.split('.access.')[0]
    twin = path.with_name(path.name.replace('.access.', '.access.json.'))
    df['source'] = 'text_gap' if twin.exists() else 'text_only'
    return df
```

```python
text_files = [p for f in [folder, folder_hd] for p in f.rglob('*.access.*log')
              if '.access.json.' not in p.name]
df_text = pd.concat([load_text(p) for p in text_files], ignore_index=True)
extra = df_text[~df_text.request_id.isin(df.request_id)].copy()
len(text_files), len(df_text), len(extra)
```

```python
TEXT_TIME = '%d/%b/%Y:%H:%M:%S %z'
extra[['method', 'uri', 'http_version']] = extra['request'].str.split(' ', n=2, expand=True) # GET /api/data HTTP/1.1
extra['datetime'] = pd.to_datetime(extra.datetime, format=TEXT_TIME).dt.tz_convert('Asia/Seoul')
for col in response_list:
    extra[col] = pd.to_numeric(extra[col], errors='coerce')
extra = extra.drop(columns=['user'])

df = pd.concat([df, extra], ignore_index=True)
df = df.replace({'': pd.NA, '-': pd.NA})
print(len(df), df.request_id.duplicated().sum())
df.source.value_counts()
```

## 8. Investigation: what is JSON missing?

```python
gap = df[df.source == 'text_gap'] # raw text files exists, in json don't this is gap
js = df[df.source == 'json'] # json files
gap.groupby(['domain', 'subdomain']).size().sort_values(ascending=False).head(8) # where it falils
```

```python
print(gap.method.value_counts())
gap.status_code.value_counts().head()
```

```python
STATIC = r'/static/|/_next/|\.(?:js|css|png|svg|woff2?|ico|jpg|webp)'
BOTS = 'bot|Uptime|curl|python'
print('static  gap:', gap.uri.str.contains(STATIC).mean().round(3),
      ' json:', js.uri.str.contains(STATIC).mean().round(3))
print('bots    gap:', gap.http_user_agent.str.contains(BOTS, case=False, na=False).mean().round(3),
      ' json:', js.http_user_agent.str.contains(BOTS, case=False, na=False).mean().round(3))
```

### Finding: JSON log skips static files
- 32,299 requests are in the text logs but not in JSON. JSON is a strict subset of text.
- 99.3% of missing rows are static assets (JS, CSS, images, fonts) vs 6.3% in JSON.
- All GET, 96% real browsers, spread across all hours. Not a time gap.
- Likely cause: an nginx `location` block for static files sets its own `access_log` with only the text format. An `access_log` inside a `location` replaces the inherited ones. Not confirmed; needs the real nginx config.
- Decision: JSON is the base, plus text rows missing from JSON (`source` = text_gap / text_only). Text rows have no country, city, proxy_ip.
- Privacy: the `user` field holds Langfuse API keys and an admin email. Dropped.
